```python
# Compute LPS Array
def compute_lps(pattern):
    m = len(pattern)
    lps = [0] * m

    length = 0
    i = 1

    while i < m:
        if pattern[i] == pattern[length]:
            length += 1
            lps[i] = length
            i += 1
        else:
            if length != 0:
                length = lps[length - 1]
            else:
                lps[i] = 0
                i += 1

    return lps


# KMP Search with Trace
def kmp_trace(text, pattern):
    lps = compute_lps(pattern)

    print("Pattern :", pattern)
    print("LPS Array:", lps)
    print()

    i = 0
    j = 0
    comparisons = 0

    while i < len(text):
        comparisons += 1

        print(f"Compare Text[{i}]='{text[i]}' with Pattern[{j}]='{pattern[j]}'")

        if text[i] == pattern[j]:
            i += 1
            j += 1

            if j == len(pattern):
                print(f"\nPattern found at index {i-j}")
                j = lps[j - 1]

        else:
            if j != 0:
                print(f"Mismatch -> Shift using LPS ({j} -> {lps[j-1]})")
                j = lps[j - 1]
            else:
                print("Mismatch -> Move to next character\n")
                i += 1

    print("\nTotal Comparisons:", comparisons)


# Main
text = "AABCAABXAAAZ"
pattern = "AABXAA"

print("Text   :", text)
print("Pattern:", pattern)
print()

kmp_trace(text, pattern)
```
