```python
import random
import time

# ---------------- Interpolation Search ----------------
def interpolation_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0

    while low <= high and arr[low] <= target <= arr[high]:
        comparisons += 1

        if low == high:
            if abs(arr[low] - target) < 1e-9:
                return low, comparisons
            return -1, comparisons

        pos = low + int(((target - arr[low]) * (high - low)) /
                        (arr[high] - arr[low]))

        if abs(arr[pos] - target) < 1e-9:
            return pos, comparisons
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1

    return -1, comparisons


# ---------------- Binary Search ----------------
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0

    while low <= high:
        comparisons += 1

        mid = (low + high) // 2

        if abs(arr[mid] - target) < 1e-9:
            return mid, comparisons
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1, comparisons


# ---------------- Performance Analysis ----------------
sizes = [10000, 50000, 100000]

print(f"{'Size':>10} {'IS Time(ms)':>14} {'BS Time(ms)':>14} {'IS Comparisons':>16} {'BS Comparisons':>16}")
print("-" * 75)

for size in sizes:

    # Uniformly distributed floating point numbers
    arr = sorted(random.uniform(0.0, 1000.0) for _ in range(size))

    target = random.choice(arr)

    # Interpolation Search Timing
    start = time.perf_counter()
    for _ in range(100):
        idx_i, comp_i = interpolation_search(arr, target)
    is_time = (time.perf_counter() - start) / 100 * 1000

    # Binary Search Timing
    start = time.perf_counter()
    for _ in range(100):
        idx_b, comp_b = binary_search(arr, target)
    bs_time = (time.perf_counter() - start) / 100 * 1000

    print(f"{size:>10} {is_time:>14.4f} {bs_time:>14.4f} {comp_i:>16} {comp_b:>16}")
```
