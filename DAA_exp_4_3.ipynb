```python
import heapq


def dijkstra_hops(graph, source):
    n = len(graph)

    dist = [float('inf')] * n
    hops = [float('inf')] * n
    prev = [None] * n

    dist[source] = 0
    hops[source] = 0

    pq = [(0, 0, source)]   # (distance, hops, vertex)

    while pq:

        d, h, u = heapq.heappop(pq)

        if d > dist[u]:
            continue

        for v, w in graph[u]:

            new_dist = dist[u] + w
            new_hops = hops[u] + 1

            if (new_dist < dist[v]) or \
               (new_dist == dist[v] and new_hops < hops[v]):

                dist[v] = new_dist
                hops[v] = new_hops
                prev[v] = u

                heapq.heappush(pq, (new_dist, new_hops, v))

    return dist, hops, prev


def path(prev, target):
    ans = []

    while target is not None:
        ans.append(target)
        target = prev[target]

    return ans[::-1]


graph = {
    0: [(1,4), (2,2)],
    1: [(3,3), (4,5)],
    2: [(3,5), (5,6)],
    3: [(4,2)],
    4: [(5,1)],
    5: []
}

dist, hops, prev = dijkstra_hops(graph, 0)

print(f"{'Vertex':>8} {'Distance':>10} {'Hops':>8} {'Path':>25}")
print("-"*60)

for i in range(len(graph)):
    p = path(prev, i)

    print(f"{i:>8} {dist[i]:>10} {hops[i]:>8} {' -> '.join(map(str,p)):>25}")
```
