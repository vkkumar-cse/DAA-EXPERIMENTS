```python
import random
import time


def prim_matrix(graph):
    n = len(graph)

    selected = [False] * n
    key = [float('inf')] * n
    parent = [-1] * n

    key[0] = 0

    for _ in range(n):

        u = -1
        minimum = float('inf')

        for i in range(n):
            if not selected[i] and key[i] < minimum:
                minimum = key[i]
                u = i

        selected[u] = True

        for v in range(n):
            if graph[u][v] != 0 and not selected[v] and graph[u][v] < key[v]:
                key[v] = graph[u][v]
                parent[v] = u

    mst = []
    cost = 0

    for i in range(1, n):
        mst.append((parent[i], i, graph[parent[i]][i]))
        cost += graph[parent[i]][i]

    return mst, cost


# -------- Dense Graph (100 vertices) --------

V = 100

graph = [[0]*V for _ in range(V)]

for i in range(V):
    for j in range(i+1, V):
        w = random.randint(1, 100)
        graph[i][j] = w
        graph[j][i] = w


start = time.perf_counter()

mst, cost = prim_matrix(graph)

end = time.perf_counter()

print("Prim's Algorithm using Adjacency Matrix")
print("Vertices :", V)
print("Total MST Cost :", cost)
print("Execution Time :", (end-start)*1000, "ms")
```
