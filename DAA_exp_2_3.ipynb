```python
import random

ROWS = 10
COLS = 10

# Generate random 10x10 matrix
text = [[random.randint(0, 9) for _ in range(COLS)] for _ in range(ROWS)]

# Choose a random 3x3 position to insert pattern
r = random.randint(0, 7)
c = random.randint(0, 7)

pattern = [
    [7, 8, 9],
    [4, 5, 6],
    [1, 2, 3]
]

# Insert pattern into text matrix
for i in range(3):
    for j in range(3):
        text[r+i][c+j] = pattern[i][j]


# Naive 2D Pattern Matching
def naive_2d_search(text, pattern):
    rows = len(text)
    cols = len(text[0])

    pr = len(pattern)
    pc = len(pattern[0])

    comparisons = 0
    matches = []

    for i in range(rows - pr + 1):
        for j in range(cols - pc + 1):

            found = True

            for x in range(pr):
                for y in range(pc):
                    comparisons += 1

                    if text[i+x][j+y] != pattern[x][y]:
                        found = False
                        break

                if not found:
                    break

            if found:
                matches.append((i, j))

    return matches, comparisons


# Display Matrix
print("10 x 10 Text Matrix\n")

for row in text:
    print(row)

print("\n3 x 3 Pattern\n")

for row in pattern:
    print(row)

matches, comparisons = naive_2d_search(text, pattern)

print("\nPattern Found At:")
for pos in matches:
    print(pos)

print("\nTotal Element Comparisons:", comparisons)
```
