```python
import random

# ---------------- Interpolation Search ----------------
def interpolation_search(arr, target):
    low, high = 0, len(arr) - 1
    probes = 0

    while low <= high and arr[low] <= target <= arr[high]:
        probes += 1

        if low == high:
            if arr[low] == target:
                return low, probes
            return -1, probes

        pos = low + int(((target - arr[low]) * (high - low)) /
                        (arr[high] - arr[low]))

        if arr[pos] == target:
            return pos, probes
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1

    return -1, probes


# ---------------- Binary Search ----------------
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    probes = 0

    while low <= high:
        probes += 1
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid, probes
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1, probes


# ---------------- Main Program ----------------
# Generate unique student roll numbers (1 to 10000)
roll_numbers = sorted(random.sample(range(1, 10001), 1000))

target = random.choice(roll_numbers)

idx_i, probe_i = interpolation_search(roll_numbers, target)
idx_b, probe_b = binary_search(roll_numbers, target)

print("Student Roll Numbers (First 20):", roll_numbers[:20], "...")
print("Searching Roll Number:", target)

print("\nInterpolation Search")
print("Position :", idx_i)
print("Probes   :", probe_i)

print("\nBinary Search")
print("Position :", idx_b)
print("Probes   :", probe_b)
```
